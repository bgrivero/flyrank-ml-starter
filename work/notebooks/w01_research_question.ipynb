## 3. Quick look at the data (2-3 real numbers)

*Load the starter CSV below and show 2-3 real numbers that make your lane look worth the next 7 weeks.*

```python
import pandas as pd

df = pd.read_csv("data/raw/content_refresh_anonymized.csv")

```

```python
df.groupby("word_count_tier")["engagement_rate"].mean()
```

```python
df["is_declining"] = (df["trend_direction"].str.lower() == "down").astype(int)
df.groupby("freshness_tier")["is_declining"].mean()
```

Checking three signals, one looked like it might actually matter: position tier seems to track with CTR pretty clearly, with top pages average roughly 10x the clicks of buried ones. The other two didn't show much of a pattern in this quick look; freshness didn't seem to relate to decline in any obvious way, and longer content didn't look tied to better engagement either. While this is early and does not suffice to make a conclusion, it shows that not every "intuitive" relationship is anchored in reality, making the track worthwhile to explore.
